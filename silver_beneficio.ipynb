#Carga dados Silver : beneficio

```python
df = spark.sql("select * from unimed.bronze.bronze_beneficio")
```

```python
(df).limit(5).display()
```

```python
#df.printSchema()
```

```python
df_tratado = df.withColumnRenamed("NU_BENEFICIO", "beneficio") \
.withColumnRenamed("NOMPCD", "nome_procedimento") \
.withColumnRenamed("INDTIPPCD", "tipo_beneficio") 
```

```python
#(df_tratado).limit(5).display()
```

```python
df_tratado = df_tratado.drop('CODTIPTAB', 'NU_FORMATO_BENEFICIO','NOME_SUBGRUPO_BENEFICIO','NOME_GRUPO_BENEFICIO','IDANEXO')
```

```python
(df_tratado).limit(5).display()
```

```python
df_tratado.write.mode("overwrite").saveAsTable("unimed.silver.silver_beneficio")
```

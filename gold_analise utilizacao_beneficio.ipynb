```python
df = spark.sql("select * from unimed.silver.silver_utilizacao_consolidada")
```

```python
df.printSchema()
```

```python
df_tratado = df.drop("dt_envio_pagamento","CODPLNCOP","NUMAUOPCDMDC")
df_tratado = df_tratado.withColumn("QTEITE", df_tratado["QTEITE"].try_cast("int"))
df_tratado = df_tratado.withColumn("VALITE", df_tratado["VALITE"].try_cast("float"))
```

```python
df_tratado.printSchema()
```

```python
df_tratado = df_tratado.withColumnsRenamed({"DE_MOTIVO_ATENDIMENTO": "motivo_atendimento",
                                             "QTEITE": "quantidade",
                                             "VALITE": "valor",
                                             "TIPPGT": "tipo_pagamento",
                                             "ID_SITUACAO": "situacao",
                                             "DTAATN": "data_atendimento",
                                             "COMPETENCIA": "competencia",
                                             "GRAU_PARTICIP": "grau_participacao",
                                             "especialidade": "especialidade",
                                             "nome_beneficio": "beneficio",
                                             "anoauopcdmcdc": "ano",
                                             "nu_guia_complemento": "guia",
                                             "id_tipo_fatura": "tipo_bendficio",
                                              "NOMTIPATN": "tipo_atendimento",
                                              "CHAVEBENEFICIARIO" : "Beneficário"
                                            })
```

```python
df_tratado.limit(5).display()
```

```python
df_tratado.write.mode("overwrite").saveAsTable("unimed.gold.gold_utilizacao_beneficio")
```

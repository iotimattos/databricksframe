#Carga dados Silver : utilizacao

```python
df = spark.sql("select * from unimed.bronze.bronze_utilizacao")
```

```python
#(df).limit(5).display()
```

```python
df.printSchema()
```

```python
df_tratado = df.withColumnRenamed("CODPLNCOP", "plano") \
.withColumnRenamed("NOMANOAUOPCDMDCPCD", "ano_atendimento") \
.withColumnRenamed("ID_TIPO_FATURA", "tipo_atendimento") \
.withColumnRenamed("INDATN", "data_atendimento") 
```

```python
#(df_tratado).limit(5).display()
```

```python
df_tratado = df_tratado.drop('NUMCPFCGCCRI', 'DTALB','DTAITN','NM_MEDICO_PROFISSIONAL','CD_PROFISSIONAL_SOLICITANTE','CBOS_SOLICITANTE','INDATN','DT_ENVIO_PAGAMENTO','VALUSUITE','CD_ESPEC_ASSOCIADA','VA_GLOSA','DE_MOTIVO_GLOSA','CODTIPTAB','TIPPGT','ID_SITUACAO','DTAATN','PARECER_AUDITOR','GRAU','VALPLNITE','VALITE','NOME_PROFISSIONAL_SOLICITANTE','QTEITE','NU_BENEFICIO','GRAU_PARTICIP','COMPETENCIA')
```

```python
(df_tratado).limit(5).display()
```

```python
#CARGA MOTIVO ATENDIMENTO
df_dimmotivoatendimento = spark.sql("select * from unimed.bronze.bronze_dimmotivoatendimento")
```

```python
df_tratado = df_tratado.join(df_dimmotivoatendimento, df_tratado.CD_MOTIVO_ATENDIMENTO == df_dimmotivoatendimento.CD_MOTIVO_ATENDIMENTO, "left")
```

```python
(df_tratado).limit(5).display()
```

```python
#CARGA tipos de ATENDIMENTO
df_tipoatendimento = spark.sql("select * from unimed.bronze.bronze_dimtipoatendimento")
df_tratado = df_tratado.join(df_tipoatendimento, df_tratado.TIPATN == df_tipoatendimento.CODTIPATN, "left")
```

```python
df_tratado = df_tratado.withColumnRenamed("CODTIPATN", "CODTIPAT") \
.withColumnRenamed("NOMANOAUOPCDMDCPCD", "ano_atendimento") 
```

```python
df_tratado = df_tratado.drop('cd_motivo_atendimento','tipatn','CODTIPAT')
```

```python
df_tratado.write.mode("overwrite").saveAsTable("unimed.silver.silver_utilizacao")
```

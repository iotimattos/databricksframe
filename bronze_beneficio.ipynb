#Craga de dados camanda bronze - Beneficios

```python
#  variaveis 
bronze_path = 'unimed.bronze.bronze_beneficio'
resource_path = '/Volumes/unimed/default/resource/origem/beneficio/*.txt'
```

```python
df_beneficio = spark.read.format('csv')\
    .option('header','true')\
    .option('delimiter',';')\
    .option("encoding", "iso-8859-1") \
    .load(resource_path)
```

```python
display(df_beneficio)
```

```python
df_beneficio.printSchema()
```

```python
# Salvar como tabela Delta (use mode overwrite se quiser substituir dados existentes)
df_beneficio.write \
    .format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable('unimed.bronze.bronze_beneficio')
```

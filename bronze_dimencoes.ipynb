#Craga de dados camanda bronze - Dimencoes

###Carga dimencao especialidade médica

```python
resource_path='/Volumes/unimed/default/resource/origem/dimencoes/GSMiEspecialidade.txt'
df_esp= spark.read.format('csv')\
    .option('header','true')\
    .option('delimiter',';')\
    .option("encoding", "iso-8859-1") \
    .load(resource_path)
```

```python
display(df_esp)
```

```python
df_esp.printSchema()
```

```python
# Salvar como tabela Delta (use mode overwrite se quiser substituir dados existentes)
df_esp.write \
    .format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable('unimed.bronze.bronze_dimespecialidade')
```

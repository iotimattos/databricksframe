#Craga de dados camanda bronze - Item Utilizacao

```python
#  variaveis 
bronze_path = 'unimed.bronze.bronze_utilizacaoitem'
resource_path = '/Volumes/unimed/default/resource/origem/utilizacaoitem/*.txt'
```

```python
df_itens = spark.read.format('csv')\
    .option('header','true')\
    .option('delimiter',';')\
    .option("encoding", "iso-8859-1") \
    .load(resource_path)
```

```python
display(df_itens)
```

```python
df_itens.printSchema()
```

```python
# Salvar como tabela Delta (use mode overwrite se quiser substituir dados existentes)
df_itens.write \
    .format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable('unimed.bronze.bronze_utilizacao')
```

#Carga dados Silver : Itens de utilizacao

```python
df = spark.sql("select * from unimed.bronze.bronze_utilizacaoitem")
```

```python
(df).limit(5).display()
```

```python
df.printSchema()
```

```python
df_tratado = df.withColumnRenamed("NU_BENEFICIO", "beneficio") \
.withColumnRenamed("NU_GUIA_COMPLEMENTO", "guia_complemento") 
```

```python
#(df_tratado).limit(5).display()
```

```python
df_tratado = df_tratado.drop('NUMCPFCGCCRI', 'VALUSUITE','VALPLNITE','VA_GLOSA','DE_MOTIVO_GLOSA','CODTIPTAB','PARECER_AUDITOR','CODEMPPAN','DTAPEVLAN','CODPLN','CD_TIPO_EMPRESTIMO','CD_TP_GER_FINAN','NM_TP_GER_FINAN','INDTIPPLAN','VALLAN','MESCOMPETENCIA')
```

```python
(df_tratado).limit(5).display()
```

```python
#CARGA especialidade
df_esp = spark.sql("select * from unimed.bronze.bronze_dimespecialidade")
```

```python
df_tratado = df_tratado.join(df_esp, df_tratado.CD_TIPO_ESPECIALIDADE == df_esp.CD_TIPO_ESPECIALIDADE, "left")
(df_tratado).limit(5).display()
```

```python
df_tratado = df_tratado.withColumnRenamed("de_tipo_especialidade", "especialidade")
df_tratado = df_tratado.drop('CD_TIPO_ESPECIALIDADE','CD_ESPEC_ASSOCIADA')
```

```python
df_tratado.write.mode("overwrite").saveAsTable("unimed.silver.silver_utilizacao_item")
```

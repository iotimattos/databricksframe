#Carga dados Silver : Utilizacao consolidada( utilizacao, utilizacao_item e beneficios)

```python
df = spark.sql("select * from unimed.silver.silver_utilizacao_item")
```

```python
(df).limit(5).display()
```

```python
#CARGA beneficio
df_beneficio = spark.sql("select beneficio,nome_procedimento as nome_beneficio from unimed.silver.silver_beneficio")
```

```python
df_tratado = df.join(df_beneficio, df.beneficio == df_beneficio.beneficio, "left")  
df_tratado = df_tratado.drop("chavebeneficiario")
(df_tratado).limit(5).display()
```

```python
#CARGA utilizacao
df_utilizacao= spark.sql("select *  from unimed.silver.silver_utilizacao")
```

```python
(df_utilizacao).limit(5).display()
```

```python
df_tratado = df_tratado.join(df_utilizacao, df_tratado.NUMFAT == df_utilizacao.NUMFAT, "left")
```

```python
(df_tratado).limit(5).display()
```

```python
df_tratado=df_tratado.drop("beneficio","NUMFAT")
```

```python
df_tratado.write.mode("overwrite").saveAsTable("unimed.silver.silver_utilizacao_consolidada")
```
